importing dependencies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
from sklearn import metrics
```

data collection and processing

```python
# loading the data from csv file to a pandas dataframe
calories = pd.read_csv('/content/calories.csv')
```

```python
# print first 5 rows of the dataframe
calories.head()
```

```python
exercise_data = pd.read_csv('/content/exercise.csv')
```

```python
exercise_data.head()
```

combining the two dataframe

```python
calories_data = pd.concat([exercise_data, calories['Calories']], axis=1)
```

```python
calories_data.head()
```

```python
# checking the number of rows and columns
calories_data.shape
```

```python
# getting some information about the dataframe
calories_data.info()
```

```python
# checking for missing values
calories_data.isnull().sum()
```

data analysis

```python
# get some stastistical measures about the data
calories_data.describe()
```

data visualisation

```python
sns.set() #provide some grid lines background theme etc
```

```python
# plotting the gender column in count plot
# since it contain male and female this is categorical column
# for categorical column we use countplot
sns.countplot(calories_data['Gender'])
```

finding the correlation in the dataset

1. positive correlation
2.negative correlation

```python
calories_data_processed = calories_data.copy()
calories_data_processed['Gender'] = calories_data_processed['Gender'].map({'female': 0, 'male': 1})

# Drop 'User_ID' as it is an identifier and not relevant for correlation analysis
calories_data_processed = calories_data_processed.drop('User_ID', axis=1)

correlation = calories_data_processed.corr()
```

```python
#constructing a heatmap to understand the correlation

plt.figure(figsize=(10,10))
sns.heatmap(correlation, cbar=True, square=True, fmt='.1f',annot=True, annot_kws={'size':18}, cmap='Blues' )
```

converting the text data to numerical values

```python
calories_data.replace({"Gender":{'male':0, 'female':1}}, inplace=True)
```

```python
calories_data.head() # now we can see that gender have been visible
```

seperating features and targets

```python
X = calories_data.drop(columns=['User_ID','Calories'], axis=1)
Y = calories_data['Calories']
```

```python
print(X)
```

```python
print(Y)
```

Splitting the data into training data and testing data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

Model Training

XGBoost Regressor

```python
# loading the model
model = XGBRegressor()
```

```python
# training the model
model.fit(X_train, Y_train)
```

Evaluation of the model

prediction on the test

```python
test_data_prediction = model.predict(X_test)
```

```python
print(test_data_prediction)
```

Mean Absolute Error

```python
mae = metrics.mean_absolute_error(Y_test, test_data_prediction)
```

```python
print("mean absolute error",mae)
```
